```python
class Person:
    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name

    def display(self, formal=False):
        if formal:
            return f"{self.last_name}, {self.first_name}"
        else:
            return f"{self.first_name}"

    def __str__(self):
        return f"Person(first_name='{self.first_name}', last_name='{self.last_name}')"
```

```python
person1 = Person("John", "Doe")

print("Informal display:", person1.display())
print("Formal display:", person1.display(formal=True))
print("__str__ representation:", str(person1))

person2 = Person("Alice", "Smith")
print("\nInformal display:", person2.display())
print("Formal display:", person2.display(formal=True))
```
