```python
ticketPrice = 10
defaultAudience = 20

print("Ad$.  Profit")
for adMoney in range(0,201,25):
  revenue = (defaultAudience + 2 * round(adMoney ** 0.5)) * ticketPrice
  costs = 200 + adMoney
  profit = revenue - costs
  print(f"{float(adMoney)} {float(profit)}")
```
