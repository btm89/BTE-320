```python
print("Hour Fee")
hour = 1
while hour < 9:
  fee = 5 + 2.5 * hour
  if fee < 10.0:
    fee = 10.0
  elif fee > 20:
    fee = 20

  hour += 1

  print(f"{hour}        {fee}")
```

```python
import math

ticketPrice = 10
defaultAudience = 20
adMoney = 0
print("Ad$\tProfit")
while adMoney <= 200:
  revenue = (defaultAudience + 2 * round(math.sqrt(adMoney))) * ticketPrice
  costs = 200 + adMoney
  profit = revenue - costs
  print(f"{float(adMoney)}\t{float(profit)}")
  adMoney += 25
```
