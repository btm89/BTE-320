```python
def reverse_string_recursive(s):
    if len(s) < 2:
        return s
    else:
        #return s[-1] + reverse(s[:-1])
        return s[-1] + reverse_string_recursive(s[1:-1]) + s[0] #Swap first and last

reverse_string_recursive('hello')
```
