```python
num1 = float(input("Enter first value: "))
num2 = float(input("Enter second value: "))
op = input("Numerical Operation (e.g., +, -, etc.): ")

if op == "+":
  print(num1 + num2)
elif op == "-":
  res = num1 - num2
elif op == "*":
  res = num1 * num2
elif op == "/":
  res = num1 / num2
elif op == "//":
  res = num1 // num2
elif op == "**":
  res = num1 ** num2
elif op == "%":
  rs = num1 % num2
else:
  print("Invalid Operation")

print(f"Result: {res}")
```
