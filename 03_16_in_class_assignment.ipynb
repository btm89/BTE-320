```python
def calcFee(hour):
  fee=5+2.5*hour
  if fee<10:
   fee=10
  elif fee > 20:
   fee=20
  print(f"Total fee:{fee}")

def main():
  hour=int(input("Enter number of hours parked: "))
  calcFee(hour)

main()
```
